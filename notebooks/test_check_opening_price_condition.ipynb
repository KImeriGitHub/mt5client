# Test Check Opening Price Condition
Run unit tests for check_opening_price_condition.py

```python
import sys
import os
from pathlib import Path

# Add project root to path
project_dir = os.path.abspath("..")
if project_dir not in sys.path:
    sys.path.append(project_dir)

print(f"Project directory: {project_dir}")
```

```python
import unittest
from tests.unit_tests.test_check_opening_price_condition import TestCheckOpeningPriceCondition
```

```python
# Create a test suite
loader = unittest.TestLoader()
suite = loader.loadTestsFromTestCase(TestCheckOpeningPriceCondition)

# Run the tests with verbose output
runner = unittest.TextTestRunner(verbosity=2)
result = runner.run(suite)
```

```python
# Print summary
print(f"\n{'='*60}")
print(f"Tests run: {result.testsRun}")
print(f"Successes: {result.testsRun - len(result.failures) - len(result.errors)}")
print(f"Failures: {len(result.failures)}")
print(f"Errors: {len(result.errors)}")
print(f"{'='*60}")
```
