```python
import sys
import os
import MetaTrader5 as mt5
import time

project_dir = os.path.abspath("..")
if project_dir not in sys.path:
    sys.path.append(project_dir)

import polars as pl

from src.infra.mtBase import mtBase
```

```python
mtb = mtBase(
    account="mt5demo_acc_usd",
    credentials_path=os.path.join("..", "secrets", "mt5_acc_cred.yaml"),
    config_path=os.path.join("..", "secrets", "mt5_config.ini"),
)
mtb.mt5_init()
```

```python
from datetime import datetime

symbols = ["KO", "INTC", "EURUSD", "TSLA", "MSFT", "USDCAD"]

for symbol in symbols:
    tick = mtb.get_symbol_price(symbol, wait_sec=0.2)
    tick_datetime = datetime.fromtimestamp(tick['time_msc'] / 1000)
    print(f"{symbol}: Ask price {tick['ask']}, {tick_datetime.strftime('%Y-%m-%d %H:%M:%S')}")
```

```python
from datetime import datetime, timezone

# Get current UTC time
current_utc = datetime.now(timezone.utc)
print(f"Current UTC time: {current_utc}")
print(f"Formatted UTC time: {current_utc.strftime('%Y-%m-%d %H:%M:%S UTC')}")

# Get current UTC timestamp in milliseconds (for comparison with tick data)
current_utc_ms = int(current_utc.timestamp() * 1000)
print(f"Current UTC in milliseconds: {current_utc_ms}")
```
